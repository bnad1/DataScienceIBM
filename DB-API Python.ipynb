```python
import sqlite3
```

```python
conn=sqlite3.connect('INSTRUCTOR.db')
```

```python
cursor_obj=conn.cursor()
```

```python
cursor_obj.execute("DROP TABLE IF EXISTS INSTRUCTOR")
```

```python
table="""CREATE TABLE IF NOT EXISTS 
      INSTRUCTOR(
      ID INTEGER PRIMARY KEY NOT NULL,
      FNAME VARCHAR(20),
      LNAME VARCHAR(20),
      CITY VARCHAR(20),
      CCODE CHAR(2)
      );
      """
```

```python
cursor_obj.execute(table)
```

```python
insert_data="""INSERT INTO INSTRUCTOR VALUES 
            (1,'Rav','Ahuja','TORONTO','CA'),
            (2,'Raul','Chong','Markham','CA'),
            (3,'Hima','Vasudevan','Chicago','US');
            """
```

```python
cursor_obj.execute(insert_data)
```

```python
# ispis red po red,svih redaka tablice INSTRUCTOR
for row in cursor_obj.execute("""SELECT * FROM INSTRUCTOR""").fetchall():
  print(row)
```

```python
# promjena grada u MOOSETOWN za osobu koja se zove Rav
cursor_obj.execute("""UPDATE INSTRUCTOR SET city='MOOSETOWN' WHERE FNAME='Rav'""")
```
